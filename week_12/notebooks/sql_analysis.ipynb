```python
import pandas as pd
import sqlite3
```

```python
telco_data = pd.read_csv("../data/raw/WA_Fn-UseC_-Telco-Customer-Churn.csv")
print(telco_data.shape)
```

```python
connection = sqlite3.connect("../data/processed/customer_analytics.db")
print("database connected")
```

```python
telco_data.to_sql("customers", connection,if_exists="replace", index=False)
print("Customer data loaded into customer table")
```

#### Total number of customers

```python
query = """
SELECT COUNT(*) AS total_customers
FROM customers;
"""
result = pd.read_sql_query(query, connection)
result
```

#### Churned Customers

```python
query = """
SELECT *
FROM customers
WHERE Churn = 'Yes';
"""

pd.read_sql_query(query, connection)
```

#### Customers who did not churn

```python
query = """
SELECT *
FROM customers
WHERE Churn = 'No';
"""

pd.read_sql_query(query, connection)
```

#### Month-to-month customers

```python
query = """
SELECT *
FROM customers
WHERE Contract = 'Month-to-month';
"""

pd.read_sql_query(query, connection)
```

#### High monthly charges

```python
query = """
SELECT customerID, tenure, MonthlyCharges, Churn
FROM customers
WHERE MonthlyCharges > 100;
"""

pd.read_sql_query(query, connection)
```

#### Fiber optic customers

```python
query = """
SELECT customerID, InternetService, MonthlyCharges, Churn
FROM customers
WHERE InternetService = 'Fiber optic';
"""

pd.read_sql_query(query, connection)
```

#### Churned + high monthly charges

```python
query = """
SELECT customerID, tenure, MonthlyCharges, Contract, Churn
FROM customers
WHERE Churn = 'Yes'
AND MonthlyCharges > 100;
"""

pd.read_sql_query(query, connection)
```

#### Average monthly charges

```python
query = """
SELECT AVG(MonthlyCharges) AS average_monthly_charges
FROM customers;
"""

pd.read_sql_query(query, connection)
```

#### Churn by contract

```python
query = """
SELECT Contract, Churn, COUNT(*) AS customer_count
FROM customers
GROUP BY Contract, Churn
ORDER BY Contract;
"""

pd.read_sql_query(query, connection)
```

#### Churn By Payment Method

```python
query = """
SELECT PaymentMethod, Churn, COUNT(*) AS customer_count
FROM customers
GROUP BY PaymentMethod, Churn
ORDER BY PaymentMethod;
"""

pd.read_sql_query(query, connection)
```

#### Average Total Charges

```python
query = """
SELECT AVG(CAST(TotalCharges AS REAL)) AS average_total_charges
FROM customers;
"""

pd.read_sql_query(query, connection)
```

#### Long-tenure customers

```python
query = """
SELECT customerID, tenure, Contract, MonthlyCharges, Churn
FROM customers
WHERE tenure >= 60
ORDER BY tenure DESC;
"""

pd.read_sql_query(query, connection)
```

#### Customer segments based on services

```python
query = """
SELECT InternetService, Contract, COUNT(*) AS customer_count
FROM customers
GROUP BY InternetService, Contract
ORDER BY customer_count DESC;
"""

pd.read_sql_query(query, connection)
```

#### Revenue-related analysis

```python
query = """
SELECT Contract, COUNT(*) AS customer_count, SUM(MonthlyCharges) AS total_monthly_revenue, AVG(MonthlyCharges) AS average_monthly_revenue
FROM customers
GROUP BY Contract
ORDER BY total_monthly_revenue DESC;
"""

pd.read_sql_query(query, connection)
```
