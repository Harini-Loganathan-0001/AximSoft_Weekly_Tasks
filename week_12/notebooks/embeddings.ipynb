```python
import pandas as pd
import numpy as np

from sentence_transformers import SentenceTransformer
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
nlp_data = pd.read_csv("../data/processed/customer_support_nlp_final.csv")
```

```python
print(nlp_data.shape)
print(nlp_data.columns.tolist())
```

```python
print("Label distribution:")
print(nlp_data["label"].value_counts())

print("\nLabel percentages:")
print((nlp_data["label"].value_counts(normalize=True) * 100).round(2))
```

```python
print("Embedding model:", model)
print("Embedding dimension:", model.get_sentence_embedding_dimension())
```

```python
# generate sentence embeddings

texts = nlp_data["text"].tolist()

embeddings = model.encode(
    texts,
    batch_size=32,
    show_progress_bar=True,
    convert_to_numpy=True
)

print(embeddings.shape)
```

```python
np.save("../data/processed/customer_support_embeddings.npy", embeddings)
print("saved")
```

```python
embedding_metadata = nlp_data.copy()
embedding_metadata.to_csv("../data/processed/customer_support_embedding_metadata.csv", index=False)

print(embedding_metadata.shape)
```
