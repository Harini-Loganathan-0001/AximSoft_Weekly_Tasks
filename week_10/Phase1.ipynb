```python
time_diff = PJME_data["Datetime"].diff()
time_diff.value_counts().head(10)
```

```python
# abnormal consumption periods
Q1 = PJME_data["PJME_MW"].quantile(0.25)
Q3 = PJME_data["PJME_MW"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = PJME_data[
    (PJME_data["PJME_MW"] < lower) |
    (PJME_data["PJME_MW"] > upper)
]

print(outliers.shape)
outliers.head()
```

```python
print("Duplicate Timestamps:", PJME_data["Datetime"].duplicated().sum())
```

```python
expected = pd.date_range(
    start=PJME_data["Datetime"].min(),
    end=PJME_data["Datetime"].max(),
    freq="h"
)

missing = expected.difference(PJME_data["Datetime"])

print("Missing Timestamps:", len(missing))
```

```python
# time seriesplot
plt.figure(figsize=(18,6))

plt.plot(
    PJME_data["Datetime"],
    PJME_data["PJME_MW"],
    color="blue",
    linewidth=0.7
)

plt.title("Hourly Electricity Demand")
plt.xlabel("Date")
plt.ylabel("Demand (MW)")
plt.savefig("images/hourly_electricity.png")
plt.show()
```

```python
# daily pattern
PJME_data["Hour"] = PJME_data["Datetime"].dt.hour

hourly = PJME_data.groupby("Hour")["PJME_MW"].mean()

plt.figure(figsize=(10,5))

sns.lineplot(
    x=hourly.index,
    y=hourly.values,
    marker="o"
)

plt.title("Average Hourly Electricity Demand")
plt.xlabel("Hour")
plt.ylabel("Average Demand")
plt.savefig("images/hourly_avg_demand.png")
plt.show()
```

```python
# weekly pattern
PJME_data["DayName"] = PJME_data["Datetime"].dt.day_name()

order = [
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
]

weekly = PJME_data.groupby("DayName")["PJME_MW"].mean().reindex(order)

plt.figure(figsize=(10,5))

sns.barplot(
    x=weekly.index,
    y=weekly.values
)

plt.title("Weekly Electricity Consumption")
plt.savefig("images/weekly_electricity_consumption.png")
plt.show()
```

```python
# monthly trend
PJME_data["Month"] = PJME_data["Datetime"].dt.month_name()

order = [
    "January","February","March","April",
    "May","June","July","August",
    "September","October","November","December"
]

monthly = PJME_data.groupby("Month")["PJME_MW"].mean().reindex(order)

plt.figure(figsize=(12,5))

sns.lineplot(
    x=monthly.index,
    y=monthly.values,
    marker="o"
)

plt.xticks(rotation=45)
plt.title("Monthly Electricity Demand")
plt.savefig("images/monthly_trend.png")
plt.show()
```

```python
# distribution plot
plt.figure(figsize=(10,5))

sns.histplot(
    PJME_data["PJME_MW"],
    bins=50,
    kde=True
)

plt.title("Demand Distribution")
plt.savefig("images/demand_distribution.png")
plt.show()
```

```python
# box plot
plt.figure(figsize=(6,6))

sns.boxplot(
    y=PJME_data["PJME_MW"]
)

plt.title("Electricity Demand Box Plot")
plt.savefig("images/demand_boxplot_electricity.png")
plt.show()
```

```python
# seasonal decomposition
from statsmodels.tsa.seasonal import seasonal_decompose

PJME_data = PJME_data.set_index("Datetime")

decomposition = seasonal_decompose(
    PJME_data["PJME_MW"],
    model="additive",
    period=24
)

decomposition.plot()
plt.savefig("images/seasonal_decomposition.png")
plt.show()
```
