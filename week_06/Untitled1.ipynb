Duplicate Record Analysis

The Geolocation dataset was examined for duplicate records. A total of 261,831 exact duplicate rows were identified, where all column values were identical. These duplicate records were removed to eliminate redundant geographic information while preserving all unique location records.

```python
geolocation.info()
```

```python
# Convert categorical

geolocation['geolocation_city'] = geolocation['geolocation_city'].astype('category')
geolocation['geolocation_state'] = geolocation['geolocation_state'].astype('category')
```

```python
# Validate Coordinates

# Latitude should be between -90 and 90
geolocation[
    (geolocation['geolocation_lat'] < -90) |
    (geolocation['geolocation_lat'] > 90)
]

# Longitude should be between -180 and 180
geolocation[
    (geolocation['geolocation_lng'] < -180) |
    (geolocation['geolocation_lng'] > 180)
]
```

```python
# Create an aggregated geolocation table

geo_agg = (
    geolocation
    .groupby('geolocation_zip_code_prefix')
    .agg({
        'geolocation_lat': 'mean',
        'geolocation_lng': 'mean',
        'geolocation_city': 'first',
        'geolocation_state': 'first'
    })
    .reset_index()
)
```

```python
geo_agg.info()
geo_agg.head()
```

```python
geo_agg['geolocation_zip_code_prefix'].duplicated().sum()
```

Geolocation Aggregation

The Geolocation dataset contains multiple coordinate records for the same ZIP code prefix. To prevent row duplication during dataset integration, the data was aggregated by geolocation_zip_code_prefix. The mean latitude and longitude were calculated, while the first available city and state values were retained. This produced a unique geographic reference for each ZIP code, enabling a one-to-one merge with the Customers and Sellers datasets.

## Missing Value Analysis After Data Preprocessing

After completing the data preprocessing phase, a missing value analysis was performed to evaluate the data quality and identify remaining incomplete records.

The remaining missing values are mainly related to order delivery details, customer reviews, and geographical information.

### Observations:

- **Order Approval Date (`order_approved_at`)**
  - A small number of records are missing approval timestamps.
  - This may occur due to incomplete payment processing records.
  - Since the missing count is very low, these records were retained.

- **Delivery Information (`order_delivered_carrier_date`, `order_delivered_customer_date`)**
  - Missing delivery dates occur because some orders were cancelled, unavailable, or not successfully completed.
  - These missing values represent real business scenarios, so they were not removed.

- **Review Information (`review_id`, `review_creation_date`, `review_answer_timestamp`)**
  - Missing review data indicates that some customers did not provide feedback after their purchase.
  - These records were preserved because the absence of reviews is meaningful customer behavior.

- **Customer and Seller Location Data**
  - Missing latitude, longitude, and geographic mapping fields occurred because some addresses could not be converted into coordinates.
  - The percentage of missing geographic records is very small, so they were retained.

### Final Decision:

The remaining missing values were not removed because they represent valid business conditions rather than data errors. Since the missing proportion is very low compared to the total dataset size, the processed dataset is suitable for further Exploratory Data Analysis (EDA).

**Conclusion:**  
The dataset quality is acceptable, and the remaining missing values will not significantly affect the reliability of business insights.

```python
# ===========================================
# Repeat Customer Analysis
# ===========================================

import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# ---------------------------------------------------
# Customer Order Analysis
# ---------------------------------------------------

customer_orders = (
    merged_df
    .groupby("customer_unique_id")["order_id"]
    .nunique()
    .reset_index()
)

customer_orders.columns = [
    "Customer_ID",
    "Order_Count"
]

# Customer Spending
customer_spending = (
    merged_df
    .groupby("customer_unique_id")["total_order_value"]
    .sum()
    .reset_index()
)

customer_spending.columns = [
    "Customer_ID",
    "Total_Spending"
]

# Merge both datasets
customer_data = pd.merge(
    customer_orders,
    customer_spending,
    on="Customer_ID"
)

# Customer Type
customer_data["Customer_Type"] = customer_data["Order_Count"].apply(
    lambda x: "Repeat Customer" if x > 1 else "One-Time Customer"
)

# ---------------------------------------------------
# KPI
# ---------------------------------------------------

total_customers = len(customer_data)

repeat_customers = (
    customer_data["Customer_Type"] == "Repeat Customer"
).sum()

one_time_customers = (
    customer_data["Customer_Type"] == "One-Time Customer"
).sum()

repeat_rate = (repeat_customers / total_customers) * 100

avg_spending = customer_data["Total_Spending"].mean()

print("=" * 50)
print("CUSTOMER ANALYSIS SUMMARY")
print("=" * 50)
print(f"Total Customers     : {total_customers:,}")
print(f"Repeat Customers    : {repeat_customers:,}")
print(f"One-Time Customers  : {one_time_customers:,}")
print(f"Repeat Rate         : {repeat_rate:.2f}%")
print(f"Average Spending    : ₹{avg_spending:,.2f}")
print("=" * 50)

# =====================================================
# Plot Style
# =====================================================

# =============================================
# Professional Dark Theme
# =============================================

plt.style.use("dark_background")

sns.set_theme(
    style="darkgrid",
    rc={
        "axes.facecolor": "#1E1E2F",
        "figure.facecolor": "#121212",
        "grid.color": "#444444",
        "axes.edgecolor": "white",
        "axes.labelcolor": "white",
        "xtick.color": "white",
        "ytick.color": "white",
        "text.color": "white"
    }
)

# =====================================================
# Figure
# =====================================================

fig, axes = plt.subplots(2, 2, figsize=(18, 14))

# Figure background
fig.patch.set_facecolor("#121212")

# Individual subplot backgrounds
for ax in axes.flat:
    ax.set_facecolor("#1E1E2F")

# =====================================================
# 1. Repeat vs One-Time Customers
# =====================================================

counts = customer_data["Customer_Type"].value_counts()

axes[0,0].bar(
    counts.index,
    counts.values,
    color=["#00E5FF", "#FF6F61"],
    edgecolor="white",
    linewidth=1.5
)

axes[0,0].set_title(
    "Repeat vs One-Time Customers",
    fontsize=16,
    fontweight="bold"
)

axes[0,0].set_ylabel("Number of Customers")

for i, value in enumerate(counts.values):
    axes[0,0].text(
        i,
        value,
        f"{value:,}",
        ha="center",
        fontsize=12,
        fontweight="bold"
    )

# =====================================================
# 2. Distribution of Orders per Customer
# =====================================================

sns.histplot(
    customer_data["Order_Count"],
    bins=20,
    kde=True,
    color="#FFD54F",
    ax=axes[0,1]
)

axes[0,1].set_title(
    "Orders per Customer",
    fontsize=16,
    fontweight="bold"
)

axes[0,1].set_xlabel("Number of Orders")

# =====================================================
# 3. Top 20 Repeat Customers
# =====================================================

top_repeat = (
    customer_data
    .sort_values("Order_Count", ascending=False)
    .head(20)
)

sns.barplot(
    data=top_repeat,
    y="Customer_ID",
    x="Order_Count",
    palette="viridis",
    ax=axes[1,0]
)

axes[1,0].set_title(
    "Top 20 Repeat Customers",
    fontsize=16,
    fontweight="bold"
)

axes[1,0].set_xlabel("Orders")

axes[1,0].set_ylabel("Customer ID")

# =====================================================
# 4. Customer Spending Distribution
# =====================================================

sns.boxplot(
    y=customer_data["Total_Spending"],
    color="#00E5FF",
    width=0.35,
    ax=axes[1,1]
)

axes[1,1].axhline(
    avg_spending,
    color="red",
    linestyle="--",
    linewidth=2,
    label=f"Average ₹{avg_spending:,.0f}"
)

axes[1,1].set_title(
    "Customer Spending Distribution",
    fontsize=16,
    fontweight="bold"
)

axes[1,1].set_ylabel("Total Spending (₹)")

axes[1,1].legend()

# =====================================================
# Dashboard Text
# =====================================================

fig.text(
    0.5,
    0.02,
    f"Total Customers: {total_customers:,}    |    "
    f"Repeat Customers: {repeat_customers:,} ({repeat_rate:.2f}%)    |    "
    f"Average Spending: ₹{avg_spending:,.0f}",
    ha="center",
    fontsize=13,
    fontweight="bold",
    color="white"
)

# =====================================================
# Final Touches
# =====================================================

for ax in axes.flat:
    ax.grid(alpha=0.25)
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)

plt.suptitle(
    "Customer & Repeat Customer Analysis Dashboard",
    fontsize=22,
    fontweight="bold",
    y=0.98
)

plt.tight_layout(rect=[0, 0.05, 1, 0.96])

# =====================================================
# Save Image
# =====================================================

plt.savefig(
    "olist_task/static/images/repeat_customer_analysis_dashboard.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()
```

```python
# Outlier Detection Report
```

```python
numeric_features = [
    "price",
    "total_order_value",
    "delivery_time",
    "shipping_duration",
    "customer_lifetime_value"
]


outlier_report = []


for col in numeric_features:

    Q1 = merged_df[col].quantile(0.25)
    Q3 = merged_df[col].quantile(0.75)

    IQR = Q3-Q1

    lower = Q1 - 1.5*IQR
    upper = Q3 + 1.5*IQR


    outliers = (
        merged_df[
            (merged_df[col]<lower) |
            (merged_df[col]>upper)
        ]
        .shape[0]
    )


    outlier_report.append(
        [
            col,
            outliers,
            round((outliers/len(merged_df))*100,2)
        ]
    )


outlier_df = pd.DataFrame(
    outlier_report,
    columns=[
        "Feature",
        "Outliers Detected",
        "Percentage"
    ]
)


outlier_df
```
